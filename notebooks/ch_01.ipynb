## Verifying Continuity by Monte Carlo

For $\Omega=[0,1]$ with Lebesgue measure $\P$ and the increasing sequence $A_n=[0,1-1/n]$, we have $A=\bigcup A_n=[0,1)$ and $\P(A)=1$. By Theorem 1.10, $\P(A_n)=1-1/n\to 1$.

*Source file:* `code/ch01/c01_1_continuity_monte_carlo.py`

```python
import numpy as np

rng = np.random.default_rng(42)
N_SAMPLES = 500_000
omega_samples = rng.uniform(0.0, 1.0, size=N_SAMPLES)

for n in [1, 5, 10, 25, 50]:
    threshold = 1.0 - (1.0 / n)
    p_sim = np.mean(omega_samples <= threshold)
    print(f"n = {n:2d} | Theoretical P(A_n): {threshold:.4f} | Monte Carlo: {p_sim:.4f}")
```

## The Birthday Problem: Exact Value and Simulation

*Source file:* `code/ch01/c01_2_birthday_problem.py`

```python
import numpy as np

rng = np.random.default_rng(42)
def exact(n):
    return 1.0 - np.prod((365 - np.arange(n)) / 365)

R = 100_000
for n in [10, 23, 30, 50]:
    days = rng.integers(0, 365, size=(R, n))
    days.sort(axis=1)
    collision = np.any(np.diff(days, axis=1) == 0, axis=1)
    approx = 1 - np.exp(-n * (n - 1) / 730)
    print(f"n={n:2d} | exact={exact(n):.4f} | exponential approx.={approx:.4f} | MC={collision.mean():.4f}")
```

## Bayes' Formula: Simulating a Diagnostic Test

*Source file:* `code/ch01/c01_3_bayes_diagnostic_test.py`

```python
import numpy as np

rng = np.random.default_rng(42)
N = 2_000_000
ill = rng.random(N) < 0.01
positive = np.where(ill, rng.random(N) < 0.99, rng.random(N) < 0.05)
print(f"P(D | +) theory    : {0.0099/0.0594:.4f}")
print(f"P(D | +) simulation: {ill[positive].mean():.4f}")
print(f"P(+)     theory    : {0.0594:.4f} | simulation: {positive.mean():.4f}")
```
